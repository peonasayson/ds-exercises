**PURPOSE:**
The purpose of prediction in this dataset is to use students' attendance records and other related details to estimate their likely grades. This helps us to identify the patterns between attendance and their performance, allowing early detection of students who may be at risk of failing. The datasets initially had many inconsistencies, errors, unnecessary characters, and missing values.


**DATA PROCESSING**

1. I removed unnecessary characters such as ., \/, and -. I used strip to remove slashes since they were at the beginning and end of a string, while for the dot and dash, I used replace since they were in the middle of a string.

2. I fixed misspelled words using replace, such as changing “fail” to “F” and “pass” to “P”.

3. I converted all attendance statuses to lowercase so I could correct their misspellings. For example, “absnt” became “absent”, “early left” became “absent”, and “late” also became “absent”.

4. After that, I capitalized the first letter of each word using str.title.

5. I arranged the subjects in alphabetical order to make them clearer and easier to read.

6. I dropped the rows in the student grade column that had “N/A”.

7. I used df.count() to check how many records were left.

8. Lastly, I reset the index so the numbering was correct.

```python
#import library
import pandas as pd
pd.set_option('display.max_rows', 110)
from google.colab import drive
drive.mount('/content/drive')
```

```python
#import dataset
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/Datasets/attendance.csv')
df
```

```python
df["Student_ID"] = df["Student_ID"].str.replace(".", "")
df["Student_ID"] = df["Student_ID"].str.replace("-", "")
df["Student_ID"] = df["Student_ID"].str.strip("\/")


df
```

```python
df["Student_Grade"] = df["Student_Grade"].str.replace("Pass", "P").replace("Fail", "F").replace("fail", "F")
df
```

```python
df["Attendance_Status"] = df["Attendance_Status"].str.lower()

df["Attendance_Status"] = df["Attendance_Status"].str.replace("left early", "absent").replace("late", "absent").replace("absnt", "absent")
df
```

```python
df["Attendance_Status"] = df["Attendance_Status"].str.title()
df
```

```python
df = df.sort_values(by='Subject', ascending=True).reset_index(drop=True)
df
```

```python
# drop empty cell
for x in df.index:
    if pd.isna(df.loc[x, "Student_Grade"]):
        df.drop(x, inplace=True)
df
```

```python
df.count()
```

```python
df = df.reset_index(drop=True)
df
```
